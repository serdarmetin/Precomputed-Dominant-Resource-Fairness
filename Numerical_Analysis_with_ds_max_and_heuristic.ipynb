```python
def pdrf():
    global c_pdrf
    for i, demand in enumerate(d_pdrf[0]):
        u_pdrf[d_pdrf[1][i]] += demand * np.floor(rates[0][i] * cycleIterations)
        c_pdrf += demand * np.floor(rates[0][i] * cycleIterations)
        s_pdrf[d_pdrf[1][i]] += ds_pdrf[0][i] * np.floor(rates[0][i] * cycleIterations)
```

```python
def heuristic():
    global c_pdrf
    for i in np.argsort(s_pdrf):
        if((r - c_pdrf >= d[i]).all()):
            u_pdrf[i] += d[i]
            c_pdrf += d[i]
            s_pdrf[i] += ds_pdrf[0][i]        
```

```python
def drf():
    global c_drf
    while(1):
        i = np.argmin(s_drf)
        if (r < d_drf[i] + c_drf).any():
            break
        else:
            u_drf[i] += d_drf[i]
            s_drf[i] += ds_drf[0][i]
            c_drf += d_drf[i]
```
